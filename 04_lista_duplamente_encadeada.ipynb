# Lista Duplamente Encadeada, Pilha e Fila

Implementação de uma **lista duplamente encadeada** (com ponteiros para o nó anterior e o próximo) e, a partir dela:

- **Pilha (LIFO):** insere e remove pelo final.
- **Fila (FIFO):** insere pelo final e remove pelo início.

Todas as operações de inserção e remoção nas pontas são O(1).

```python
class DoubleLinkedList:

    class Node:
        def __init__(self, element):
            self.element = element
            self.next = None
            self.prev = None

    def __init__(self):
        self.head = None
        self.tail = None

    def insert_head(self, element):
        new_node = self.Node(element)
        if self.head is None:
            self.head = new_node
            self.tail = new_node
        else:
            new_node.next = self.head
            self.head.prev = new_node
            self.head = new_node

    def remove_head(self):
        if self.head is None:
            return None
        element = self.head.element
        self.head = self.head.next
        if self.head is not None:
            self.head.prev = None
        else:
            self.tail = None
        return element

    def insert_tail(self, element):
        new_node = self.Node(element)
        if self.tail is None:
            self.head = new_node
            self.tail = new_node
        else:
            new_node.prev = self.tail
            self.tail.next = new_node
            self.tail = new_node

    def remove_tail(self):
        if self.tail is None:
            return None
        element = self.tail.element
        self.tail = self.tail.prev
        if self.tail is not None:
            self.tail.next = None
        else:
            self.head = None
        return element

    def print(self):
        current = self.head
        while current is not None:
            print(current.element, end=" ")
            current = current.next
        print()


class Pilha:
    def __init__(self):
        self.lista = DoubleLinkedList()

    def push(self, elemento):
        self.lista.insert_tail(elemento)

    def pop(self):
        return self.lista.remove_tail()

    def topo(self):
        return None if self.lista.tail is None else self.lista.tail.element

    def vazia(self):
        return self.lista.head is None

    def mostrar(self):
        self.lista.print()


class Fila:
    def __init__(self):
        self.lista = DoubleLinkedList()

    def enqueue(self, elemento):
        self.lista.insert_tail(elemento)

    def dequeue(self):
        return self.lista.remove_head()

    def primeiro(self):
        return None if self.lista.head is None else self.lista.head.element

    def vazia(self):
        return self.lista.head is None

    def mostrar(self):
        self.lista.print()

print("=== Pilha ===")
p = Pilha()
p.push(10)
p.push(20)
p.push(30)
p.mostrar()
print("Pop:", p.pop())
print("Topo:", p.topo())
p.mostrar()

print("\n=== Fila ===")
f = Fila()
f.enqueue("A")
f.enqueue("B")
f.enqueue("C")
f.mostrar()
print("Dequeue:", f.dequeue())
print("Primeiro:", f.primeiro())
f.mostrar()
```
