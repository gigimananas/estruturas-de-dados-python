# Validador de HTML com Pilha

Implementação de uma **pilha** (LIFO) em Python e aplicação prática: verificar se as tags de um documento HTML abrem e fecham na ordem correta.

- Tags de abertura são empilhadas; cada tag de fechamento deve corresponder ao topo da pilha.
- Tags auto-fechadas (`<br/>`, `<img/>`) são ignoradas e atributos (`<img src=...>`) são desconsiderados na comparação.
- Ao final, o HTML só é válido se a pilha estiver vazia.

```python
# Classe Html
class Html:
  def __init__(self):
    self.dados = []

  def size(self):
    return len(self.dados)

  def push(self, e):
    self.dados.append(e)

  def pop(self):
    if self.is_empty():
      return None
    ret = self.dados[-1]
    del self.dados[-1]
    return ret

  def top(self):
    return self.dados[-1]

  def is_empty(self):
    return self.size() == 0
```

```python
# Função para extrair todas as tags <...>
def extrair_tags(html):
    tags = []
    i = 0
    while i < len(html):
        if html[i] == "<":
            j = i
            while j < len(html) and html[j] != ">":
                j += 1
            tags.append(html[i:j+1])  # pega a tag completa
            i = j
        i += 1
    return tags


# Função para verificar HTML
def verificar_html(html):
    pilha = Html()
    tags = extrair_tags(html)

    for tag in tags:
        if tag.endswith("/>"):
            # tag auto-fechada (ex: <br/>, <img/>), ignora
            continue
        elif not tag.startswith("</"):
            # tag de abertura normal
            pilha.push(tag)
        else:
            # tag de fechamento
            if pilha.is_empty():
                return False
            abertura = pilha.pop()
            # compara nomes das tags (tirando <, > e /)
            nome_abertura = abertura.strip("<>").split()[0]  # ignora atributos
            nome_fechamento = tag.strip("</>")
            if nome_abertura != nome_fechamento:
                return False

    return pilha.is_empty()

html_1 = """
<body>
<center>
<h1> O barquinho </h1>
</center>
<p> A tempestade jogou o pequeno
barco como um tênis barato em um
máquina de lavar velha, Os três
pescadores bébados estavam acostumados
tal tratamento, é claro, mas
não o vendedor de árvores, que mesmo
um clandestino agora sentiu que ele
pagou a mais pela viagem. </p>
<ol>
<li> O vendedor morrerá? </li>
<li> Qual é a cor do barco? </li>
<li> E a Noemi? </li>
</ol>
</body>
"""

html_2 = """
<html>
  <head>
    <title>Página de Teste</title>
  </head>
  <body>
    <h1>Título</h1>
    <p>Parágrafo <strong>com negrito</strong> e <em>itálico</em>.</p>
  </body>
</html>
"""

html_3 = """
<html>
  <body>
    <div>
      <h2>Lista de Itens</h2>
      <ul>
        <li>Item A</li>
        <li>Item B</li>
        <li>Item C</li>
      </ul>
    </div>
  </body>
</html>
"""

html_4 = """
<html>
  <body>
    <h1>Galeria</h1>
    <img src="foto1.png"/>
    <img src="foto2.png"/>
    <br/>
    <p>Fim da galeria.</p>
  </body>
</html>
"""

html_5 = """
<html>
  <body>
    <h1>Título</h2>
    <p>Texto aqui</p>
  </body>
</html>
"""

html_6 = """
<html>
  <head>
    <title>Teste</title>
  </head>
  <body>
    <div>
      <p>Parágrafo sem fechamento
  </body>
</html>
"""

html_7 = """
<html>
  <body>
    <b><i>Texto em negrito e itálico</b></i>
  </body>
</html>
"""

# Exemplos de uso
print(verificar_html(html_1)) #True
print(verificar_html(html_2)) #True
print(verificar_html(html_3)) #True
print(verificar_html(html_4)) #True
print(verificar_html(html_5)) #False
print(verificar_html(html_6)) #False
print(verificar_html(html_7)) #False
print(verificar_html("<html><body><h1>Título</h1></body></html>"))  # True
print(verificar_html("<html><body><h1>Título</body></html>"))       # False (h1 sem fechamento)
print(verificar_html("<div><p>Texto</p><br/></div>"))               # True (br é auto-fechada)
print(verificar_html("<ul><li>Item</li><li>Item 2</ul>"))           # False (li não fechado)
```
