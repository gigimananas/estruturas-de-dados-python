# Filas de Prioridade e Algoritmos de Ordenação

Duas implementações de **fila de prioridade**:

| Implementação | Inserção | Busca do mínimo | Ordenação associada |
|---|---|---|---|
| Não ordenada | O(1) | O(n) | Selection Sort |
| Ordenada | O(n) | O(1) | Insertion Sort |

O notebook mostra a troca de custo entre inserir e buscar, e implementa os dois algoritmos de ordenação.

```python
class PriorityQueue_Unsorted:
  class Item:
    def __init__(self, key, value):
      self.key = key
      self.value = value

  def __init__(self):
    self.data = []

  def len(self): return len(self.data)
  def is_empty(self): return len(self.data) == 0

  def add(self, key, value):
    self.data.append(self.Item(key, value))

  def min(self):
    if self.is_empty(): return None
    min_item = self.data[0]
    for item in self.data:
      if item.key < min_item.key:
        min_item = item
    return (min_item)

  def remove_min(self):
    min_item = self.min()
    if min_item == None: return None
    else: self.data.remove(self.min())
    return min_item

  def print(self):
    rep = [f"({item.key}, {item.value})" for item in self.data]
    res = "{" + ",".join(rep) + "}"
    print(res)

  # Selection Sort
  def selection_sort(self, ulist):
    n = len(ulist)
    #percorre a lista inteira
    for i in range(n):
        #assume que o menor elemento está no resto da lista
        min_index = i
        #procura o menor elemento no resto da lista
        for j in range(i + 1, n):
            if ulist[j] < ulist[min_index]:
                min_index = j #guarda índice do novo menor
        #remove o menor encontrado da lista
        min_value = ulist.pop(min_index)

        # insere o menor valor na posição correta(i)
        ulist.insert(i, min_value)
    return ulist


class PriorityQueue_Sorted:
  class Item:
    def __init__(self, key, value):
      self.key = key
      self.value = value

  def __init__(self):
    self.data = []

  def len(self): return len(self.data)
  def is_empty(self): return len(self.data) == 0

  def add(self, key, value):
    new_item = self.Item(key, value)
    if self.is_empty() or new_item.key < self.data[0].key:
      self.data.insert(0, new_item)
    else:
      inserted = False
      for item in self.data:
        if new_item.key < item.key:
          self.data.insert(self.data.index(item), new_item)
          inserted = True
          break
      if not inserted:
        self.data.append(new_item)

  def min(self):
    if self.is_empty(): return None
    min_item = self.data[0]
    return (min_item)

  def remove_min(self):
    min_item = self.min()
    if min_item == None: return None
    else: self.data.remove(self.min())
    return min_item

  def print(self):
    rep = [f"({item.key}, {item.value})" for item in self.data]
    res = "{" + ",".join(rep) + "}"
    print(res)

  # Insertion Sort
  def insertion_sort(self, ulist):
    #percorre a lista a partir do 2º elemento
    for i in range(1, len(ulist)):
        #pega o elemento que será colocado no lugar certo
        current = ulist.pop(i)
        #compara com o elemento anterior
        j = i - 1
        #enquanto não chegamos no início e o elemento da esquerda for maior
        while j >= 0 and ulist[j] > current:
            j -= 1 #anda uma posição para a esquerda
        #insere o elemento atual na posição correta encontrada
        ulist.insert(j + 1, current)
    return ulist



lista = [7, 3, -1, 13, 27, 0, 12, 222, 342, -500, 1]

pq_uns = PriorityQueue_Unsorted()
print("Selection Sort:", pq_uns.selection_sort(lista.copy()))

pq_srt = PriorityQueue_Sorted()
print("Insertion Sort:", pq_srt.insertion_sort(lista.copy()))

# Usando as filas de prioridade (chave = prioridade, valor = tarefa)
for pq in (PriorityQueue_Unsorted(), PriorityQueue_Sorted()):
  pq.add(3, "relatório")
  pq.add(1, "urgente")
  pq.add(2, "e-mail")
  pq.print()
  item = pq.remove_min()
  print("Removido:", item.key, item.value)
  pq.print()
```
